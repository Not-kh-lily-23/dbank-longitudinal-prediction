```python
import re
import pandas as pd
import pylangacq
import os
data='/content/drive/MyDrive/DementiaBank Project'
drive='/content/drive/MyDrive/DementiaBank Project'
def chat_mdata(fpath):
    try:
        fname=os.path.basename(fpath)
        f_id=os.path.splitext(fname)[0]
        with open(fpath, 'r', encoding='utf-8',errors='ignore') as f:
            content=f.read()
        id_line=re.search(r'@ID:\s*.*',content)
        id_str=id_line.group(0) if id_line else ""
        field=id_str.split('|')
        grp=field[5] if len(field)>5 else "Unknown"
        age=field[3] if len(field)>3 else None
        edu=field[8] if len(field)>8 else None
        mmse=field[9] if len(field)>9 else None
        return {
            "file_id": f_id,
            "file_path": fpath,
            "age": age,
            "group": grp,
            "education": edu,
            "mmse": mmse
        }
    except Exception as e:
        print(f"Error on {fpath}: {e}")
        return None
record=[]
for root,_,files in os.walk(data):
    for file in files:
        if file.endswith('.cha'):
            full_path=os.path.join(root,file)
            p_data=chat_mdata(full_path)
            if p_data:
                record.append(p_data)
df=pd.DataFrame(record)
op_path=os.path.join(drive,'pitt_corpus_content.csv')
df.to_csv(op_path,index=False)
print(f"Successfully parsed {len(df)} transcripts.")
print(f"Manifest saved to: {op_path}")
df.head(10)
```
