```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
from google.colab import drive
drive.mount('/content/drive')
base='/content/drive/MyDrive/DementiaBank Project'
csv_path=os.path.join(base,'nlp_features_final.csv')
df=pd.read_csv(csv_path)
features=['mlu_w','mattr','pn_ratio','disfluency_rate']
fig,axes=plt.subplots(2,2,figsize=(14,10))
axes=axes.flatten()
for i, feat in enumerate(features):
    sns.boxplot(x='cohort_status',y=feat,data=df,ax=axes[i],order=['Stable Control','Stable MCI','Stable AD'],palette='Set2')
    axes[i].set_title(f'Distribution of {feat.upper()}')
    axes[i].set_ylabel('')
    axes[i].set_xlabel('')
plt.tight_layout()
viz_path=os.path.join(base,'Figure2_Distributions.png')
plt.savefig(viz_path,dpi=300)
print(f"Figure 2 saved to: {viz_path}")
plt.show()
```
